```python
# 1.1 identify base sensor columns
curr_cols  = [c for c in df_clean.columns if c.startswith('Current_J')]
speed_cols = [c for c in df_clean.columns if c.startswith('Speed_J')]

# 1.2 direction & magnitude
for col in curr_cols + speed_cols:
    df_clean[f'{col}_abs']  = df_clean[col].abs()
    df_clean[f'{col}_sign'] = np.sign(df_clean[col])

# 1.3 joint power (Current × Speed)
for j in range(6):
    df_clean[f'Power_J{j}'] = df_clean[f'Current_J{j}'] * df_clean[f'Speed_J{j}']

# 1.4 aggregate loads
df_clean['TotalCurrent'] = df_clean[curr_cols].sum(axis=1)
df_clean['TotalSpeed']   = df_clean[speed_cols].sum(axis=1)

# 1.5 rolling mean temperature (30-sample window)
df_clean['Temp_T0_ma30'] = (
    df_clean['Temperature_T0']
      .rolling(30, min_periods=1)
      .mean()
)

# 1.6 warm-up & stress flags
df_clean['is_warmup']  = (df_clean['Temperature_T0'] < 35).astype(int)
df_clean['stress_flag'] = (
    df_clean[curr_cols].abs().gt(3).any(axis=1).astype(int)
)
print("Finished feature engineering → total columns:", len(df_clean.columns))
```

#### Feature Engineering/selection Anlaysis
- Power_J0–Power_J5 (Current × Speed): Capture joint-level work done; highly predictive of tool effort.
- TotalCurrent & TotalSpeed: Summarize overall electrical and movement load on the robot.
- Current_Jk_abs / Speed_Jk_abs: Capture magnitude regardless of direction — useful when load matters more than motion direction.
- Current_Jk_sign / Speed_Jk_sign: Encode motion direction separately, allowing the model to distinguish task phases (e.g., grip vs release).
- Temp_T0_ma30: Smoothed temperature trend, adds stability and captures heating effects over time.
- is_warmup: Flags early-stage operation (cold start), helps model adjust baseline.
- stress_flag: Flags high-load snapshots; isolates extreme cases affecting tool strain.
These features improve model interpretability and performance without introducing complexity.

```python
# Prepare the x and y
target   = "Tool_current"
exclude  = ["Timestamp", "Num", "cycle", target]

X = (df_clean
     .drop(columns=[c for c in exclude if c in df_clean.columns])
     .select_dtypes(exclude="object"))
y = df_clean[target]
```

```python
# ============================================================
# 1.  Train / test split
# ============================================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42)

# ============================================================
# 2.  Gradient-Boosting Regressor
# ============================================================
gbr = GradientBoostingRegressor(
    n_estimators=400,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)
gbr.fit(X_train, y_train)

y_pred = gbr.predict(X_test)
print(f"Test R²  = {r2_score(y_test, y_pred):.3f}")
print(f"Test MAE = {mean_absolute_error(y_test, y_pred):.4f} A")
```

```python
# ============================================================
# 3.  Save model & feature list for real-time use
# ============================================================
model_path = "tool_current_gbr.joblib"
joblib.dump({"model": gbr, "features": X.columns.tolist()}, model_path)
print(f"✔️  Model saved ➜ {model_path}")
# ============================================================
# 4.  Real-time inference helper
# ============================================================
def predict_tool_current(sensor_row: dict) -> float:
    """
    Predict Tool_current (amps) from a single sensor snapshot.

    sensor_row: dict with keys matching engineered feature names
                (Current_J*, Speed_J*, Power_J*, TotalCurrent, etc.)
    """
    bundle = joblib.load(model_path)
    feats  = bundle["features"]
    model  = bundle["model"]

    # Build 1-row DataFrame, fill missing engineered columns with 0
    row_df = pd.DataFrame([sensor_row])
    for col in feats:
        if col not in row_df.columns:
            row_df[col] = 0.0
    row_df = row_df[feats]            # reorder
    return float(model.predict(row_df)[0])

# ---------- quick demo ----------
demo_row = X_test.iloc[0].to_dict()
print("Predicted Tool_current =", predict_tool_current(demo_row), "A",
      "\nActual Tool_current    =", y_test.iloc[0], "A")
```

```python
# Predict test set
import seaborn as sns
y_pred = gbr.predict(X_test)

# Create scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x=y_test, y=y_pred, alpha=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()],
         color='black', linestyle='--', label='Perfect prediction')

plt.xlabel("Actual Tool_current")
plt.ylabel("Predicted Tool_current")
plt.title(f"Gradient Boosting Regression\nTest R² = {r2_score(y_test, y_pred):.3f} | MAE = {mean_absolute_error(y_test, y_pred):.4f}")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# --- 1. Load the Dataset ---
# Make sure 'ur3_cobotops.csv' is in the same directory as your Python script
try:
    df = pd.read_csv("UR3_CobotOps-cleaned.csv")
    print("Dataset loaded successfully.")
    print(f"Dataset shape: {df.shape}")
    print("First 5 rows of the dataset:")
    print(df.head())
except FileNotFoundError:
    print("Error: 'ur3_cobotops.csv' not found.")
    print("Please download the UR3 CobotOps dataset from UCI ML Repository and place it in the script's directory.")
    exit() # Exit if the file isn't found

# --- 2. Data Preprocessing and Feature Selection ---

# Let's predict the temperature of Joint 0 (Temperature_J0)
# We'll use relevant features like its own current and speed,
# and potentially other joint currents/speeds as they can influence overall robot heat.
# T0 is a reference temperature sensor, which could also be a good predictor.

# Define features (X) and target (y)
# We select features that are likely to influence Temperature_J0
features = [
    'Current_J0', 'Speed_J0', # Joint 0's own current and speed
    'Current_J1', 'Speed_J1', # Current and speed of adjacent joints
    'Current_J2', 'Speed_J2',
    'Current_J3', 'Speed_J3',
    'Current_J4', 'Speed_J4',
    'Current_J5', 'Speed_J5'
]
target = 'Temperature_J0'

# Ensure all selected features and the target exist in the DataFrame
missing_cols = [col for col in features + [target] if col not in df.columns]
if missing_cols:
    print(f"\nError: Missing columns in the dataset: {missing_cols}")
    print("Please check the column names or the dataset integrity.")
    exit()

X = df[features]
y = df[target]

# Handle potential missing values in selected columns (simple imputation for demonstration)
# In a real-world scenario, you might use more sophisticated imputation methods.
X = X.fillna(X.mean())
y = y.fillna(y.mean())

print(f"\nFeatures selected: {features}")
print(f"Target selected: {target}")
print(f"Shape of X: {X.shape}, Shape of y: {y.shape}")

# --- 3. Split Data into Training and Testing Sets ---
# For time-series data, it's generally better to split chronologically to avoid data leakage.
# We'll set shuffle=False to maintain the temporal order.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False, random_state=42)

print(f"\nData split into training ({len(X_train)} samples) and testing ({len(X_test)} samples).")

# --- 4. Train the Linear Regression Model ---
model = LinearRegression()
model.fit(X_train, y_train)

print("\nLinear Regression Model Trained.")
print(f"Model Intercept: {model.intercept_:.2f}")

# Display coefficients for each feature
print("\nModel Coefficients:")
for feature, coef in zip(features, model.coef_):
    print(f"  {feature}: {coef:.4f}")

# --- 5. Evaluate the Model ---
y_pred = model.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse) # Root Mean Squared Error
r2 = r2_score(y_test, y_pred)

print(f"\nModel Evaluation on Test Set:")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R-squared (R2): {r2:.4f}") # R2 close to 1 indicates a good fit

# --- 6. Visualize Results ---
plt.figure(figsize=(12, 7))

# Plot actual vs. predicted values
plt.subplot(1, 2, 1)
plt.scatter(y_test, y_pred, alpha=0.5)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', lw=2, label='Perfect Prediction') # A line showing where perfect predictions would lie
plt.xlabel(f"Actual {target}")
plt.ylabel(f"Predicted {target}")
plt.title(f"Actual vs. Predicted {target}")
plt.grid(True)
plt.legend()

# Plot residuals
residuals = y_test - y_pred
plt.subplot(1, 2, 2)
plt.scatter(y_pred, residuals, alpha=0.5)
plt.axhline(y=0, color='r', linestyle='--', lw=2)
plt.xlabel(f"Predicted {target}")
plt.ylabel("Residuals (Actual - Predicted)")
plt.title("Residual Plot")
plt.grid(True)

plt.tight_layout()
plt.show()

# Example: Make a prediction for a hypothetical new data point
# Create a dummy new data point with values for all features
# (You'd replace these with actual live sensor readings)
# Ensure the order of features matches 'features' list above
new_data_point_values = [
    5.0, 1.0,  # Current_J0, Speed_J0
    2.0, 0.5,  # Current_J1, Speed_J1
    3.0, 0.8,  # Current_J2, Speed_J2
    1.5, 0.3,  # Current_J3, Speed_J3
    2.5, 0.7,  # Current_J4, Speed_J4
    1.0, 0.2,  # Current_J5, Speed_J5
    25.0      # T0
]
new_data_point = pd.DataFrame([new_data_point_values], columns=features)
predicted_temp = model.predict(new_data_point)

print(f"\nPrediction for a new hypothetical data point:")
print(f"Input features: {new_data_point_values}")
print(f"Predicted {target}: {predicted_temp[0]:.2f} °C")
```
